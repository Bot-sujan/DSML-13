```python
user_profile = {
    "username": "johndoe",
    "age": 30,
    "email": "johndoe@example.com",
    "skills": ["Python", "JavaScript", "SQL"]
}

marks = {"math": 85, "science": 90, "history": 78}

new_dict = user_profile | marks
print(new_dict)
```

NESTED Dict

```python
user_profile = {
    "username": "johndoe",
    "age": 30,
    "email": "johndoe@example.com",
    "skills": ["Python", "JavaScript", "SQL"],
    "marks": {"math": 85, "science": 90, "history": 78},
}

user_profile["marks"]["english"] = 88
print(user_profile)
```

In dict, keys are immutable and it cannot be changed.
