Python Set Type
A set is an unordered collection of unique elements in Python. Sets are mutable, meaning their contents can be changed after they are created. Sets are created using curly braces{} or the set() constructor. 

Creating a Set
To create a set, you can use curly braces{} with comma-seperated values inside or the set() constructor with a list as an arguement. Here are some examples:

```python
setA = {4, 5, 8, 1, 1, 1, 2, 3}
type(setA)
setB = {"messissippi", "missouri", "India", "United States", "Canada", "Mexico"}
type(setB)
```

Set Operations
Sets in Python support a variety of operations, including union, intersection, difference and sysmmetric difference. Here are some examples:- 

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}
c = a & b , # Intersection
d = a | b  ,# Union
e = a - b  ,# Difference
```
