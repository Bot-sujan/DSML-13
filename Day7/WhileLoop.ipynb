```python
# While program to calculate the sum of the first N whole numbers.

n = 50
add = 0
i = 1
while i <= n:
    add += i
    i += 1
print(add)
```

```python
# break statement
# if we want to break the loop at a certain point we can use break inside the loops.

i = 0
while i < 10:
    print(i)
    if i == 5:
        break
    i += 1
```

continue statement
if we want to skip loops at a certain point we can use continue insode the loops(works for both while and for loops). Here's an example.

```python
# printing fibonacci series using any loops

a = 0
b = 1
n = 10
count = 0
while count < n:

    print(a)
    c = a + b
    a = b
    b = c
    count += 1
```
