Python Comprehension
In python, comprehension provide a consice way to create sequences like lists, dictionaries, and sets. They are more readable and often more efficent alternative using traditional loops to generate sequence. 

There are three types of comprehensions in Python: list comprehensions, dictionary comprehensions and set comprehensions. 

# List Comprehension
syntax = 
    output = [<expression> for items in iterables if <condition>]
    

```python
squared = [i ** 2 for i in range(1, 11) if i % 2 == 0]
squared
```

List comprehension can also be nested, allowing you to create more complex sentences. Here is an example:

```python
i = 0
while i < 2:
    i += 1
    j = 0
    while j < 5:
        j += 1
        if j == 3:
            break
        print(i, j)
```

```python
chars = ['c', 'a', 'b']
pairs = { char1 + char2 for char1 in chars for char2 in chars }
pairs
```
