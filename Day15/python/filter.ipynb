```python
# filter prime numbers from the given list. 


def is_prime(n):
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True


test_numbers = [
    -10,
    -1,
    0,
    1,
    2,
    3,
    4,
    5,
    6,
    7,
    9,
    11,
    13,
    15,
    17,
    19,
    21,
    25,
    27,
    47,
    49,
    79,
    97,
    100,
    523,
    527,
    1000,
    7919,
    10000,
    104729,
    1000000,
]

# Filter prime numbers using list comprehension
prime_numbers = [num for num in test_numbers if is_prime(num)]

print(prime_numbers)
```
