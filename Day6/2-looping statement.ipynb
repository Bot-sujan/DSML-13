Python Loop Statements
In python, loops are used to execute a block of code repeatedly. There are two types of loops in Python: for loops and while loops. 

PYTHON FOR LOOP:
  syntax:
     for variable in sequence:
            do something

The item in a nre variable that is created with each iteration og the loop, and takes on a value of the next element in the items. 

```python
fruits =  ["apple", "banana", "cherry"]

for fruit in fruits:
    if fruit == "banana":
        print("I found a banana!")
        break
    print(fruit)
```

```python
for i in range(1,11):
    print(i)
```

```python
base_str = "*"
for i in range(1,6):
    print(base_str * i)
```

Nested for loops

```python
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    for ad in adj:
        print(ad, fruit)
```

```python
# multiplication table of a number 

num = int(input("Enter a number: "))
# num = 5
for i in range(1, 11):
    print(f"{num} x {i} = {num*i}")
```

Multiple arguements in Loop.

```python
person = {
    "name": "John",
    "age": 30,
    "surName": "Doe",
    "city": "New York"
}

for item in person.items():
    print(item)
```

```python
coordinate = [(1, 2), (3, 4), (5, 6)]
for x, y, z in coordinate:
    print(x, y, z)
```
