IF condition
The condition is an expression that returns a boolean value(True or False)

```python
score = 30
if score >= 30:
    print("You passed!")
    print("Inside if block")
```

```python
num = 77

if num % 2 == 0:
    print(f"{num} is even")
else:
    print(f"{num} is odd")
```

```python
x = "madam"

if x == x[::-1]:
    print(f"{x} is a palindrome")
else:
    print(f"{x} is not a palindrome")
```

In addition to the else clause, the if statement can also be followed by elif clause. 

      if: 
        condition1
      elif:
        condition2
      else:
        condition3

```python
# marks to grade. 

marks = 69.70
if marks >= 90:
    print("A+")
elif marks >= 80 and marks < 90:
    print("A")
elif marks >= 70  and marks < 80:
    print("B")
elif marks >= 60  and marks < 70: 
    print("C")
else:
    print("Fail")
```
