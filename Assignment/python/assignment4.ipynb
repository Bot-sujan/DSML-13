Math Tutor Using Random, Arithmetic Operators, and OOP

Scenario: Create a MathTutor class that generates random math questions using random and math operators.

Track correct and incorrect answers.
Provide a score at the end.
Handle invalid input using exception handling.
Concepts: Random, Math, Exception Handling, Classes

```python
import operator
import random


class MathTutor:
    """A math tutor system that generates dynamic math questions,

    tracks user performance, and handles invalid inputs gracefully.
    """

    # Map visual operator strings to standard library arithmetic functions
    OPERATIONS = {
        "+": operator.add,
        "-": operator.sub,
        "*": operator.mul,
    }

    def __init__(self, total_questions=5):
        """Initialize the score counters and set total questions."""
        self.total_questions = total_questions
        self.correct_answers = 0
        self.total_asked = 0

    def generate_question(self):
        """Generates two random operands and a random operation.

        Returns a tuple: (question_str, correct_answer)
        """
        symbol, func = random.choice(list(self.OPERATIONS.items()))

        # Generate numbers between 1 and 20 for basic arithmetic
        num1 = random.randint(1, 20)
        num2 = random.randint(1, 20)

        # Ensure subtraction doesn't yield negative results (optional adjustment)
        if symbol == "-" and num1 < num2:
            num1, num2 = num2, num1

        answer = func(num1, num2)
        question_str = f"What is {num1} {symbol} {num2}? "

        return question_str, answer

    def get_valid_input(self, prompt):
        """Prompts the user repeatedly until a valid integer is entered.

        Handles non-numeric string entries using exception handling.
        """
        while True:
            try:
                user_input = input(prompt).strip()
                return int(user_input)
            except ValueError:
                print("⚠️ Invalid input! Please enter a whole number.")

    def run_quiz(self):
        """Main loop that conducts the quiz session."""
        print("=" * 45)
        print(f" Welcome to Math Tutor! ({self.total_questions} Questions)")
        print("=" * 45)

        for i in range(1, self.total_questions + 1):
            print(f"\nQuestion {i} of {self.total_questions}:")
            question_str, correct_answer = self.generate_question()

            # Get user response safely
            user_answer = self.get_valid_input(question_str)
            self.total_asked += 1

            # Check answer
            if user_answer == correct_answer:
                print(" Correct!")
                self.correct_answers += 1
            else:
                print(f"❌ Incorrect. The right answer was {correct_answer}.")

        self.display_score()

    def display_score(self):
        """Displays the final summary and percentage score."""
        percentage = (self.correct_answers / self.total_asked) * 100

        print("\n" + "=" * 45)
        print("                QUIZ RESULTS                ")
        print("=" * 45)
        print(f"Total Questions : {self.total_asked}")
        print(f"Correct Answers : {self.correct_answers}")
        print(f"Incorrect       : {self.total_asked - self.correct_answers}")
        print(f"Final Score     : {percentage:.1f}%")
        print("=" * 45)


# Entry point
if __name__ == "__main__":
    tutor = MathTutor(total_questions=5)
    tutor.run_quiz()
```

Exception Handling Scenario: Online Age-Restricted Service

Scenario: You’re building a sign-up system for an online movie rental platform. Some movies are age-restricted (18+). You need to ensure proper validation and error handling during user registration.

Task:

Create a custom exception class called UnderageError that inherits from Exception.

Write a function register_user() that:

Takes a user’s name and age as input.
Raises UnderageError if the user is under 18.
Otherwise, prints a welcome message.
Wrap the function call in a try block and handle the exception.

Use else to confirm successful registration and finally to always print “Thank you for using MovieTime!” regardless of outcome.

Also try to validate if the age input is numeric. Raise a ValueError if not, and handle it separately.

```python
# 1. Custom Exception Class
class UnderageError(Exception):
    """Raised when a user is under 18 years old."""

    pass


# 2. Registration Function
def register_user(name: str, age_input: str):
    # Validate numeric input
    if not str(age_input).strip().isdigit():
        raise ValueError("Age must be a valid numeric whole number.")

    age = int(age_input)

    # Validate age restriction
    if age < 18:
        raise UnderageError(
            f"Registration failed for {name}. You must be 18 or older."
        )

    print(f"🎉 Welcome to MovieTime, {name}!")


# 3. Main execution block with try-except-else-finally
def main():
    name = input("Enter your name: ").strip()
    age_input = input("Enter your age: ").strip()

    try:
        register_user(name, age_input)
    except ValueError as ve:
        print(f"❌ Input Error: {ve}")
    except UnderageError as ue:
        print(f"🚫 Age Restriction Error: {ue}")
    else:
        print("✅ Account successfully registered!")
    finally:
        print("Thank you for using MovieTime!")


if __name__ == "__main__":
    main()
```

We want to build an online shopping cart system that allows users to add products to their cart, calculate the total cost, apply discounts, and generate an invoice. The system should include the following functionalities:

Adding products to the cart
Removing products from the cart
Calculating the total cost
Applying discounts based on user type
Generating an invoice

```python
from datetime import datetime


class Product:
    """Represents an individual product in the store."""

    def __init__(self, product_id: str, name: str, price: float):
        if price < 0:
            raise ValueError("Price cannot be negative.")
        self.product_id = product_id
        self.name = name
        self.price = price


class CartItem:
    """Represents a product and its quantity inside the cart."""

    def __init__(self, product: Product, quantity: int = 1):
        if quantity <= 0:
            raise ValueError("Quantity must be greater than zero.")
        self.product = product
        self.quantity = quantity

    def get_subtotal(self) -> float:
        return self.product.price * self.quantity


class ShoppingCart:
    """Manages cart items, applies discounts based on user type, and prints invoices."""

    # Discount percentages mapped by user type
    USER_DISCOUNTS = {
        "REGULAR": 0.00,  # 0% discount
        "SILVER": 0.05,  # 5% discount
        "GOLD": 0.10,  # 10% discount
        "VIP": 0.20,  # 20% discount
    }

    def __init__(self, user_name: str, user_type: str = "REGULAR"):
        self.user_name = user_name
        self.user_type = user_type.upper()
        self.items: dict[str, CartItem] = {}  # product_id -> CartItem

        if self.user_type not in self.USER_DISCOUNTS:
            raise ValueError(
                f"Invalid user type: {user_type}. Choose from {list(self.USER_DISCOUNTS.keys())}"
            )

    def add_product(self, product: Product, quantity: int = 1):
        """Adds a product to the cart or increases quantity if it already exists."""
        if product.product_id in self.items:
            self.items[product.product_id].quantity += quantity
        else:
            self.items[product.product_id] = CartItem(product, quantity)
        print(
            f"➕ Added {quantity}x '{product.name}' to {self.user_name}'s cart."
        )

    def remove_product(self, product_id: str, quantity: int = None):
        """Removes a product or decreases its quantity in the cart."""
        if product_id not in self.items:
            print("⚠️ Item not found in cart.")
            return

        if quantity is None or quantity >= self.items[product_id].quantity:
            removed_item = self.items.pop(product_id)
            print(f"🗑️ Removed '{removed_item.product.name}' from cart.")
        else:
            self.items[product_id].quantity -= quantity
            print(
                f"➖ Reduced '{self.items[product_id].product.name}' quantity by {quantity}."
            )

    def calculate_subtotal(self) -> float:
        """Calculates total cost before discount."""
        return sum(item.get_subtotal() for item in self.items.values())

    def calculate_discount(self) -> float:
        """Calculates total discount amount based on user type."""
        subtotal = self.calculate_subtotal()
        discount_rate = self.USER_DISCOUNTS.get(self.user_type, 0.0)
        return subtotal * discount_rate

    def calculate_total(self) -> float:
        """Calculates final total after applying discount."""
        return self.calculate_subtotal() - self.calculate_discount()

    def generate_invoice(self):
        """Prints a structured itemized invoice."""
        if not self.items:
            print("\n🛒 Your shopping cart is empty!")
            return

        subtotal = self.calculate_subtotal()
        discount_amount = self.calculate_discount()
        discount_pct = int(self.USER_DISCOUNTS[self.user_type] * 100)
        final_total = self.calculate_total()

        print("\n" + "=" * 50)
        print("                OFFICIAL INVOICE                ")
        print("=" * 50)
        print(f"Customer Name : {self.user_name}")
        print(f"Customer Type : {self.user_type}")
        print(f"Date          : {datetime.now().strftime('%Y-%m-%d %H:%M')}")
        print("-" * 50)
        print(f"{'Item':<22} {'Qty':<6} {'Price':<10} {'Subtotal':<10}")
        print("-" * 50)

        for item in self.items.values():
            p = item.product
            print(
                f"{p.name:<22} {item.quantity:<6} ${p.price:<9.2f} ${item.get_subtotal():<9.2f}"
            )

        print("-" * 50)
        print(f"{'Subtotal:':<38} ${subtotal:>9.2f}")
        print(f"{f'Discount ({discount_pct}%):':<38} -${discount_amount:>8.2f}")
        print("=" * 50)
        print(f"{'TOTAL DUE:':<38} ${final_total:>9.2f}")
        print("=" * 50 + "\n")


# Demonstration
if __name__ == "__main__":
    # Create products
    laptop = Product("P001", "Gaming Laptop", 1200.00)
    mouse = Product("P002", "Wireless Mouse", 25.50)
    keyboard = Product("P003", "Mechanical Keyboard", 85.00)

    # Initialize cart for a VIP user
    cart = ShoppingCart(user_name="Sarah Connor", user_type="VIP")

    # 1. Add products
    cart.add_product(laptop, quantity=1)
    cart.add_product(mouse, quantity=2)
    cart.add_product(keyboard, quantity=1)

    # 2. Modify cart (remove 1 mouse)
    cart.remove_product("P002", quantity=1)

    # 3. Generate initial invoice
    cart.generate_invoice()
```
