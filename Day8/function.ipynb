open()
In python the open() function is used to open a file and return a file object. This file object can be used to read or write data to the file. Once the file is opened the read() function is called on the file object to read the contents of the file into a string variable, by specifying the mode as "r".

```python
file = open("welcome.txt", "w")
file.write("This is first log. \n")
file.write("This file is written by Sujan using Python. \n")
file.write("This is another log, with a detailed explaination of how \n the file is written using Python.")

file.close()

```

```python
# Context Manager
with open("welcome.txt", mode = "r+") as file:
    print(file.read())
    file.write("\n Hello all")
print("closed")
```

zip()
The zip() function in Python is used to combine two or more iterables into a single iterable that returns tuples containing corresponding elements from each input iterable. 
The syntax for zip() is"
  zip(iterable1, iterable2......)

```python
name = ["Hari", "Shyam", "Sujan", "Sandesh"]
score = [90, 80, 70, 60]
section = ["A", "B", "C", "D"]

list(zip(name, score, section))
```

```python
for a, b, c in zip(name, score, section):
    print(a, b, c)
```
