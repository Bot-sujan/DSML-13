```python
class PassCheck:
    def __init__(self, password: str):
        self.password = password

    def __call__(self) -> str:
        has_alpha = any(char.isalpha() for char in self.password)
        has_numeric = any(char.isdigit() for char in self.password)
        is_valid_length = len(self.password) >= 8

        if is_valid_length and has_alpha and has_numeric:
            return "Strong Password"
        return "Weak Password"


# Usage example:
pass_check = PassCheck("Sujan@23456")
print(pass_check())  # Output: Strong Password
```
