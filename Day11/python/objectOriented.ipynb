Calculating Student Result:
Develop a class to accept a student's name and marksin three subjects, then calculate and display the total and average marks.
total(), average(), display()

```python
class StudentResult:
    def __init__(self, name: str, mark1: float, mark2: float, mark3: float):
        self.name = name
        self.mark1 = mark1
        self.mark2 = mark2
        self.mark3 = mark3

    def total(self) -> float:
        return self.mark1 + self.mark2 + self.mark3

    def average(self) -> float:
        return self.total() / 3

    def display(self) -> None:
        print(f"Student: {self.name}")
        print(f"Marks: {self.mark1}, {self.mark2}, {self.mark3}")
        print(f"Total: {self.total()}")
        print(f"Average: {self.average():.2f}")


# Example
student = StudentResult("Aakash", 78, 85, 92)
student.display()

student = StudentResult("Sujan", 57, 29, 40)
student.display()
```
