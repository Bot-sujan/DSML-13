DICTIONARY 

Creating a dictionary

```python
myDict = {
    "name" : "Hari",
    "age" : 25,
    "city" : "Bengaluru",
    3: "marks",
    "another": {"home": "Nepal"}
}
print(myDict)
```

Modifying Elements
You can modify the value of a dictionary elements by specifying the key and assigning a new value to it. 

```python
myDict = {
    "name" : "Hari",
    "age" : 25,
    "city" : "Bengaluru",
    3: "marks",
    "another": {"home": "Nepal"}
}
print(myDict)
myDict.pop("age")
```
