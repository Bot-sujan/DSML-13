#Accessing List Items
You can access individual items in a list using indexing. In Python, indexing starts with 0. Here are some examples. 

list[start index : stop index : steps]

```python
numbers = list(range(1, 31))
print(numbers)
```

```python
odd_numbers = numbers[::2]
print(odd_numbers)
```

```python
even_numbers = numbers[1::2]
print(even_numbers)
```

```python
print(numbers[::-1])
```

Modifying Lists
You can modify individual items in a list by assigning a new value to the index. You can also add or remove items from a list using built-in-list methods. Here are some examples. 

```python
fruits = ["apple", "banana", "cherry", "date"]
fruits
```

```python
fruits.append("elderberry")
fruits
```

```python
fruits.insert(3,'mango')
fruits  
```

```python
fruits.extend(["kiwi", "grape"])
fruits
```

```python
fruits.remove("banana")
fruits
```

```python
fruits.pop(4)
fruits  
```

LIST METHODS:- 
python provides a variety of built-in list methods that you can use to perform operations on lists. Some common list methods include sort(), reverse(), count() and index(). Here are some examples:- 

```python
myList = [3, 1, 99, 15, 3.2, 0]
myList.sort(reverse=True)
myList
```

```python
myList.index(3.2)
min(myList)
max(myList)
sum(myList)
```
